```python
# 生成器函数
def gen_ab():
    print('start')
    yield 'a'
    print('continue')
    yield 'b'
    print('end.')

for c in gen_ab():
    print('-->', c)
```

## 标准库中的生成器函数

### 用于筛选的生成器函数
从输入的可迭代对象中产出项的子集，而且不修改项本身。

```python
def vowel(c):
    return c.lower() in 'aeiou'
```

```python
# filter(predicate, it)
# 把 it 中的各个元素传给 predicate，
# 如果 predicate(item) 返回真值，那么产出对应的元素；
# 如果 predicate 是 None，那么只产出真值元素
list(filter(vowel, 'Aardvark'))
```

```python
# filterfalse(predicate, it)
# 与 filter 函数的作用类似，不过 predicate的逻辑是相反的：predicate 返回假值时产出对应的项
list(itertools.filterfalse(vowel, 'Aardvark'))
```

```python
# dropwhile(predicate, it)
# 处理 it，跳过 predicate 的计算结果为真值的项，然后产出剩下的项
list(itertools.dropwhile(vowel, 'Aardvark'))
```

```python
# takewhile(predicate, it)
# predicate 返回真值时产出对应的项
list(itertools.takewhile(vowel, 'Aardvark'))
```

```python
# compress(it, selector_it)
# 并行处理两个可迭代对象；如果 selector_it中的项为真值，那么产出 it 中对应的项
list(itertools.compress('Aardvark', (1, 0, 1, 1, 0, 1)))
```

```python
# islice(it, stop) 或 islice (it, start, stop, step=1)
# 产出 it 的切片，作用类似于 s[:stop] 或 s[start:stop:step]，不过 it 可以是任何可迭代对象，而且这个函数惰性执行操作
print(list(itertools.islice('Aardvark', 4)))
print(list(itertools.islice('Aardvark', 4, 7)))
print(list(itertools.islice('Aardvark', 1, 7, 2)))
```

### 用于映射的生成器函数
在输入的可迭代对象（map 和 starmap 函数可处理多个可迭代对象）中的各项上做计算，产出计算结果。

```python
# accumulate(it,[func])
# 产出累计求和；如果提供了 func，那么把前两个项传给它，然后把计算结果和下一项传给它，以此类推，产出最后结果
sample = [5, 4, 2, 8, 7, 6, 3, 0, 9, 1]
import itertools
# 累计和
print(list(itertools.accumulate(sample)))
# 累计最小值
print(list(itertools.accumulate(sample, min)))
# 累计最大值
print(list(itertools.accumulate(sample, max)))

import operator
# 累计积
print(list(itertools.accumulate(sample, operator.mul)))
# 从 1 到 10，计算各个数的阶乘
print(list(itertools.accumulate(range(1, 11), operator.mul)))
```

```python
# yield from
def sub_gen():
    yield 1.1
    yield 1.2
    return 'Sub Done!'

def gen():
    yield 1
    result = yield from sub_gen()
    print('<--', result)
    yield 2

for x in gen():
    print(x)
```
